```python
import pandas as pd
data={
    "Name":["Amit","Riya","Vishal","Yash","Priya"],
    "Age":[20,21,22,21,22],
    "Department":["CSE","IT","CSE","ECE","IT"],
    "Marks":[85,90,78,88,76]
}

df=pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df.tail())
print("Shape:",df.shape)
print("Columns:",df.columns.tolist())
print("\n Data types:")
print(df.dtypes)
print("\n Information:")
df.info()
print("\n Statistics:")
df.describe()
```

```python
iris_df=pd.read_csv('iris.csv')
iris_df.head()
```

```python
df.to_csv('student_data.csv',index=False)
```

```python
df_new=pd.read_csv('student_data.csv')
df_new
```

```python
print("Shape:", iris_df.shape)
print("Columns:", iris_df.columns.tolist())
print("\n Data types:")
print(iris_df.dtypes)
print("\n Missing values:")
print(iris_df.isnull().sum())
print("\n Information:")
iris_df.info()
print("\n Summary:")
print(iris_df.describe(include='all'))
```
